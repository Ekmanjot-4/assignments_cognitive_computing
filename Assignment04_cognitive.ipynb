```python
df=pd.DataFrame(fixed_entries)
df
```

```python
# Question 2 - Generate and score a hypothesis
def score_query(query,df):
  query_keywords=query.lower().split()
  results=[]
  for _,row in df.iterrows():
    faq_keywords=row['keywords'].lower().split()
    score =0
    for word in query_keywords:
      if word in faq_keywords:
        score+=1
    if score>0:
      results.append((row['question'],row['answer'],score))
    results.sort(key=lambda x:x[2],reverse=True)
    return results
```

```python
query=input("Enter your query: ")
results=score_query(query,df)
for question,answer,score in results:
  print(f"Question: {question}")
  print(f"Answer: {answer}")
  print(f"Confidence Score: {score}")
```

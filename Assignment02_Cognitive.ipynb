```python
import random
random_list=[]
random.seed(1024170377)
random.randint(100,900)
for i in range(0,100):
  random_list.append(random.randint(100,900))
random_list

```

```python
odd=[]
even=[]
for x in random_list:
  if(x%2==0):
    even.append(x)
  else:
    odd.append(x)
print(f"The list for odd numbersi is : {odd}")
print(f"The list for even numbers is :{even}")
```

```python
import math
def is_prime(x):
  for i in range(2,int(math.sqrt(x)+1)):
    if (x%i==0):
      return False
  return True
```

```python
prime=[]
for x in random_list:
  if(is_prime(x)):
    prime.append(x)
print(f"The prime numbers in the list are {prime}")
```

```python
counts=[]
for x in random_list:
  counts.append(random_list.count(x))
max_count=max(counts)
index =counts.index(max_count)
most_frequent=random_list[index]
freq=[]
for i in range(len(counts)):
  if counts[i]==max_count:
    freq.append(random_list[i])
print(f"The numbers with maximum frequency in the random list are :{freq}")
print(f"The amount of times these numbers occur is {max_count}")
```

```python
# Question 4: 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
#each digit} and set B = {digit × 9 for each digit}. Print both sets.
A=set()
B=set()
roll=1024170377
for i in range(0,8):
  A.add(roll%10*7)
  B.add(roll%10*9)
  roll=roll//10
print(A)
print(B)
```

```python
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(B.difference(A))
print(A.symmetric_difference(B))
print(A.issubset(B))
print(B.issubset(A))
X=int(input("Enter a value:"))
A.discard(X)
print(A)
# discard is safer because it does not throw an error whenever input value is not present in set
```

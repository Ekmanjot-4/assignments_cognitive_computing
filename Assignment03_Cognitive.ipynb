```python
avg_salary=emp['Salary'].mean()
total_bonus=emp['Bonus'].sum()
youngest_age=emp['Age'].min()
highest_rating=emp['Rating'].max()
print("Average Salary:", avg_salary)
print("Total Bonus:", total_bonus)
print("Youngest Age:", youngest_age)
print("Highest Performance Rating:", highest_rating)
```

```python
emp=emp.sort_values(by="Salary",ascending=False)
```

```python
emp["Performance Category"]=emp["Rating"].apply(
    lambda x: "Excellent" if x>=4.5
    else "Good" if x>=4.0
    else "Average"
)
emp
```

```python
print(emp.isnull().sum())
```

```python
emp.rename(columns={"Employee_ID":"ID"},inplace=True)
emp
```

```python
print(emp[emp['Years_of_Experience']>5])
```

```python
print(emp[emp['Department']=='IT'])
```

```python
emp['Tax']=emp['Salary'].apply(
    lambda x: x-0.10*x
)
emp
```

```python
emp.to_csv("emp_result.csv",index=False)
```
